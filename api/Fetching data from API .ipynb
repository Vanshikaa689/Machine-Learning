```python
df = pd.DataFrame()
```

```python
print(df)
```

```python
for i in range(1, 15):
    response = requests.get("https://api.themoviedb.org/3/tv/airing_today?api_key=a43f4f09cfaf19caa45f2b7402fc2b3e&language=en-US&page={}".format(i))
    temp_df = pd.DataFrame(response.json()['results'])[['id', 'vote_average']]
    df = pd.concat([df, temp_df], ignore_index = True)
print(df)
```

```python
df.to_csv('movies.csv')
```
